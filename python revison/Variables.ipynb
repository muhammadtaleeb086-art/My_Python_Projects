## Variables 
A variable is a container used to store data in memory.

- Variables store data
- Python does not require datatype declaration 
- Variable names should be meaningful 
- Avoid using spaces in variable names 

```python
## Question 1: Create a variable named `student_name` and store your name in it. Print the variable.

student_name = "Mohammed Taleeb Chouhan "
print(student_name)
```
